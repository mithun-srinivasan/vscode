```python
import tensorflow as tf
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

# Import files module for file uploads in Colab
from google.colab import files

# Load pre-trained model
model = tf.keras.applications.MobileNetV2(weights="imagenet")

# Load image
# Upload image via Colab's file upload utility
uploaded = files.upload()
image_path = list(uploaded.keys())[0]
img = Image.open(image_path).resize((224, 224))

# Display image
plt.imshow(img)
plt.axis("off")

# Preprocess
img_array = np.array(img)
img_array = np.expand_dims(img_array, axis=0)
img_array = tf.keras.applications.mobilenet_v2.preprocess_input(img_array)

# Predict
predictions = model.predict(img_array)
result = tf.keras.applications.mobilenet_v2.decode_predictions(predictions, top=1)[0]

print("\nPrediction Result:")
print("Object:", result[0][1])
print("Confidence:", round(result[0][2]*100, 2), "%")
```
